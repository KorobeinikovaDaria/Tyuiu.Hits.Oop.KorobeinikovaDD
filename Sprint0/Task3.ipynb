Создать базовый класс Order в C#, который будет представлять информацию о
заказах товаров или услуг. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
// Товар
public class Item
{
    public string Name { get; set; }
    public int Amount { get; set; }
    public string Delivery { get; set; }
}

// Базовый класс
class Order
{
    // Атрибуты
    public int OrderId { get; set; }
    public string CreationDate { get; set; }
    public int TotalAmount { get; set; } 

    public List<string> names= new List<string>();
    public List<int> amounts = new List<int>();

    public virtual void CalculateTotal()
    {   
        int sum = 0;
        for (int i = 0; i < amounts.Count; i++)
        {
            sum += amounts[i];
        }
        
        TotalAmount = sum;
        Console.WriteLine($"Cумма заказа: {sum}");
    }

    // Метод для добавления элемента в заказ
    public virtual void AddItem(Item item)
    {
       amounts.Add(item.Amount);
       names.Add(item.Name);
       Console.WriteLine($"Добавлен товар '{item.Name}' в заказ");
    }

    // Метод для удаления элемента из заказа
    public virtual void RemoveItem(Item item)
    {  
       if (names.Contains(item.Name))
       { 
       amounts.Remove(item.Amount);
       names.Remove(item.Name);
       Console.WriteLine($"Удален товар '{item.Name}' из заказа");
       }
       else 
       { 
        Console.WriteLine($"Товар '{item.Name}' не найден");
       }
    }
}

// Пример
Order getOrder = new Order {OrderId = 019988, CreationDate = "03.09.25"};

Item oneItem = new Item { Name = "Тетрадь", Amount = 200};
getOrder.AddItem(oneItem);
getOrder.CalculateTotal();

Item twoItem = new Item { Name = "Потрфель", Amount = 2300};
getOrder.AddItem(twoItem);

Item threeItem = new Item { Name = "Гуашь", Amount = 800};
getOrder.AddItem(threeItem);
getOrder.CalculateTotal();

getOrder.RemoveItem(oneItem);
getOrder.CalculateTotal();
getOrder.RemoveItem(oneItem);
```

```
class onlineOrder : Order
{ 
    public string CustomerEmai { get; set; }
    List<string> deliveries = new List<string>();

    public virtual void AddItem(Item item)
    {
       amounts.Add(item.Amount);
       names.Add(item.Name);
       deliveries.Add(item.Delivery);
       Console.WriteLine($"Добавлен товар '{item.Name}' в заказ");
    }
}

// Пример
onlineOrder getOnlineOrder = new onlineOrder {OrderId = 019988, CreationDate = "03.09.25", CustomerEmai = "myordermail.ru" };

Item oneItem = new Item { Name = "Тетрадь", Amount = 200, Delivery = "Голубинная почта"};
getOnlineOrder.AddItem(oneItem);
getOnlineOrder.CalculateTotal();

Item twoItem = new Item { Name = "Потрфель", Amount = 2300, Delivery = "Почта России"};
getOnlineOrder.AddItem(twoItem);

getOnlineOrder.RemoveItem(oneItem);
getOnlineOrder.CalculateTotal();
getOnlineOrder.RemoveItem(oneItem);
```
