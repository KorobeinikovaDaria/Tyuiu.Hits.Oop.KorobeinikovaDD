### Структура и Объявление Класса в C#

В C#, класс объявляется с использованием ключевого слова `class`, за которым следует имя класса. Класс может содержать поля (данные), свойства (способ доступа к данным), методы (функциональность) и события. 

#### Свойства

Свойства в C# используются для инкапсуляции данных. Они предоставляют гибкий механизм для чтения, записи или вычисления значений приватных полей. Свойства могут иметь два акцессора: `get` (геттер) для чтения значения и `set` (сеттер) для его изменения.

#### Геттеры и Сеттеры

Геттеры и сеттеры являются частью свойств класса в C#. Геттер (`get`) используется для возврата значения свойства, а сеттер (`set`) — для установки нового значения. Сеттер может включать логику валидации или другие операции перед изменением значения поля.

### Пример Классов Vehicle и Car

Давайте рассмотрим примеры классов `Vehicle` и `Car`, чтобы проиллюстрировать использование свойств, геттеров и сеттеров, а также разницу между статическими и обычными атрибутами и методами.

#### Класс Vehicle

```csharp
public class Vehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с геттером и сеттером
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException("Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор
    public Vehicle()
    {
        TotalVehicles++;
    }

    // Обычный метод
    public void Accelerate(int delta)
    {
        Speed += delta;
    }

    // Статический метод
    public static void DisplayTotalVehicles()
    {
        Console.WriteLine($"Общее количество транспортных средств: {TotalVehicles}");
    }
}
```

#### Класс Car

```csharp
public class Car : Vehicle
{
    private int _fuelLevel;

    // Свойство FuelLevel с геттером и сеттером
    public int FuelLevel
    {
        get { return _fuelLevel; }
        set { _fuelLevel = value; }
    }

    // Конструктор
    public Car() : base()
    {
        FuelLevel = 100; // Задаем начальный уровень топлива
    }

    // Обычный метод
    public void Refuel(int amount)
    {
        FuelLevel += amount;
    }
}
```

### Разница между Статическими и Обычными Атрибутами и Методами Класса

- **Статические атрибуты и методы** принадлежат самому классу, а не его экземплярам. Они могут быть вызваны без создания объекта класса. В примере выше `TotalVehicles` является статическим атрибутом, а `DisplayTotalVehicles()` — статическим методом класса `Vehicle`.
- **Обычные атрибуты и методы** принадлежат экземплярам класса и могут иметь разные значения для разных объектов. В примере выше `_color`, `_speed` и `_fuelLevel` являются обычными атрибутами, а методы `Accelerate()` и `Refuel()` — обычными методами.

````
public class Vehicle
{
    private string _color;
    private int _speed;

    // Свойство Color с геттером и сеттером
    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    // Свойство Speed с геттером и сеттером
    public int Speed
    {
        get { return _speed; }
        set
        {
            if (value >= 0)
                _speed = value;
            else
                throw new ArgumentOutOfRangeException("Скорость не может быть отрицательной!");
        }
    }

    // Статический атрибут
    public static int TotalVehicles { get; private set; } = 0;

    // Конструктор
    public Vehicle()
    {
        TotalVehicles++;
    }

    // Обычный метод
    public void Accelerate(int delta)
    {
        Speed += delta;
    }

    // Статический метод
    public static void DisplayTotalVehicles()
    {
        Console.WriteLine($"Общее количество транспортных средств: {TotalVehicles}");
    }
}

public class Car : Vehicle
{
    private int _fuelLevel;

    // Свойство FuelLevel с геттером и сеттером
    public int FuelLevel
    {
        get { return _fuelLevel; }
        set { _fuelLevel = value; }
    }

    // Конструктор
    public Car() : base()
    {
        FuelLevel = 100; // Задаем начальный уровень топлива
    }

    // Обычный метод
    public void Refuel(int amount)
    {
        FuelLevel += amount;
    }
}
````

````
// Базовый класс
    public class Animal
    {
        // Приватные поля
        private string _name;
        private int _age;
        private double _weight;

        // Свойства с геттером и сеттером (с валидацией)
        public string Name
        {
            get => _name;
            set
            {
                if (string.IsNullOrWhiteSpace(value))
                    throw new ArgumentException("Имя не может быть пустым.");
                _name = value;
            }
        }

        public int Age
        {
            get => _age;
            set
            {
                if (value < 0)
                    throw new ArgumentOutOfRangeException(nameof(Age), "Возраст не может быть отрицательным.");
                _age = value;
            }
        }

        public double Weight
        {
            get => _weight;
            set
            {
                if (value <= 0)
                    throw new ArgumentOutOfRangeException(nameof(Weight), "Вес должен быть положительным.");
                _weight = value;
            }
        }

        // Статический атрибут — общее количество созданных животных
        public static int TotalAnimals { get; private set; } = 0;

        // Обычный атрибут только для чтения (инициализируется в конструкторе)
        public readonly DateTime CreatedAt;

        // Конструктор
        public Animal(string name, int age, double weight)
        {
            Name = name;
            Age = age;
            Weight = weight;
            CreatedAt = DateTime.Now;
            TotalAnimals++;
        }

        // Виртуальный метод, который могут переопределить потомки
        public virtual void Speak()
        {
            Console.WriteLine($"{Name} издает неопределённый звук.");
        }

        // Обычные методы
        public void Eat(double foodWeight)
        {
            if (foodWeight <= 0) throw new ArgumentOutOfRangeException(nameof(foodWeight), "Пища должна иметь положительный вес.");
            // Простая логика: прибавляем часть веса от съеденной пищи
            Weight += foodWeight * 0.1;
            Console.WriteLine($"{Name} поел(а) {foodWeight} ед., новый вес: {Weight:F2}.");
        }

        public void Move(int distanceMeters)
        {
            if (distanceMeters < 0) throw new ArgumentOutOfRangeException(nameof(distanceMeters));
            Console.WriteLine($"{Name} прошёл(ла) {distanceMeters} м.");
        }

        // Статический метод
        public static void DisplayTotalAnimals()
        {
            Console.WriteLine($"Всего животных: {TotalAnimals}");
        }
    }

    // Производный класс Dog
    public class Dog : Animal
    {
        private string _breed;

        public string Breed
        {
            get => _breed;
            set => _breed = string.IsNullOrWhiteSpace(value) ? "Unknown" : value;
        }

        // Автоматическое свойство
        public bool IsTrained { get; set; }

        // Статический атрибут, считающий собак
        public static int TotalDogs { get; private set; } = 0;

        public Dog(string name, int age, double weight, string breed, bool isTrained = false)
            : base(name, age, weight)
        {
            Breed = breed;
            IsTrained = isTrained;
            TotalDogs++;
        }

        public override void Speak()
        {
            Console.WriteLine($"{Name} (собака) лает: Гав-гав!");
        }

        public void Fetch(string item)
        {
            Console.WriteLine($"{Name} принес(ла) {item}.");
        }

        public static void DisplayTotalDogs()
        {
            Console.WriteLine($"Всего собак: {TotalDogs}");
        }
    }

    // Производный класс Cat
    public class Cat : Animal
    { 
        private bool _isIndoor;

        public bool IsIndoor
        {
            get => _isIndoor;
            set => _isIndoor = value;
        }

        // Пример статического свойства-константы (не изменяемое)
        public static readonly int DefaultLives = 9;

        public Cat(string name, int age, double weight, bool isIndoor = true)
            : base(name, age, weight)
        {
            IsIndoor = isIndoor;
        }

        public override void Speak()
        {
            Console.WriteLine($"{Name} (кот) мяукает: Мяу!");
        }

        public void Scratch()
        {
            Console.WriteLine($"{Name} точит когти.");
        }
    }

    // Производный класс Bird
    public class Bird : Animal
    {
        public double WingSpan { get; set; } // размах крыльев в сантиметрах

        public Bird(string name, int age, double weight, double wingSpan)
            : base(name, age, weight)
        {
            if (wingSpan <= 0) throw new ArgumentOutOfRangeException(nameof(wingSpan));
            WingSpan = wingSpan;
        }

        public override void Speak()
        {
            Console.WriteLine($"{Name} (птица) чирикает.");
        }

        public void Fly(int distanceMeters)
        {
            if (distanceMeters < 0) throw new ArgumentOutOfRangeException(nameof(distanceMeters));
            Console.WriteLine($"{Name} пролетел(а) {distanceMeters} м.");
        }
    }

    // Пример использования
    class Program
    {
        static void Main()
        {
            Dog d1 = new Dog("Барбос", 3, 20.5, "Овчарка", isTrained: true);
            Cat c1 = new Cat("Мурка", 2, 4.2);
            Bird b1 = new Bird("Кеша", 1, 0.3, 25.0);

            d1.Speak();
            d1.Fetch("мячик");
            d1.Eat(0.5);

            c1.Speak();
            c1.Scratch();
            c1.Move(10);

            b1.Speak();
            b1.Fly(200);

            Animal.DisplayTotalAnimals();
            Dog.DisplayTotalDogs();

            // Примеры геттеров/сеттеров и валидации
            try
            {
                c1.Age = -1; // вызовет исключение
            }
            catch (ArgumentOutOfRangeException ex)
            {
                Console.WriteLine($"Ошибка установки: {ex.Message}");
            }
        }
    }
````
