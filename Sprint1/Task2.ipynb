<h1 style="color:DodgerBlue">Создание экземпляров классов (объектов) и взаимодействие объектов между собой</h1>

### Создание экземпляров классов (объектов)

В C#, экземпляр класса создается с помощью оператора `new`, который выделяет память для нового объекта и вызывает конструктор класса для инициализации этого объекта. Конструктор — это специальный метод в классе, который имеет то же имя, что и класс, и используется для установки начального состояния объекта.

#### Пример создания экземпляров классов Vehicle и Car

```csharp
Vehicle myVehicle = new Vehicle();
Car myCar = new Car();
```

В этом примере создаются два объекта: один для класса `Vehicle` и другой для класса `Car`. Конструкторы этих классов будут автоматически вызваны при создании объектов.

### Взаимодействие объектов между Собой

Взаимодействие объектов между собой означает, что один объект может вызывать методы другого объекта или обращаться к его свойствам (если они доступны). Это позволяет объектам обмениваться информацией и координировать свои действия.

#### Пример взаимодействия объектов

```csharp
myVehicle.Color = "Red";
myVehicle.Speed = 60;

myCar.Color = "Blue";
myCar.FuelLevel = 100;
myCar.Refuel(50); // Увеличиваем уровень топлива на 50
```

В этом примере мы устанавливаем свойства `Color` и `Speed` для объекта `myVehicle`, а также свойства `Color` и `FuelLevel` для объекта `myCar`. Затем мы вызываем метод `Refuel` у объекта `myCar`, чтобы увеличить его уровень топлива.

### Закрепление правил использования модификаторов доступа

Модификаторы доступа определяют уровень доступности членов класса из других частей кода. Важно правильно использовать модификаторы доступа для обеспечения безопасности и инкапсуляции данных.

#### Примеры модификаторов доступа в классах Vehicle и Car

```csharp
public class Vehicle
{
    public string Color { get; set; } // public свойство
    private int speed; // private поле

    protected void SetSpeed(int value) // protected метод
    {
        speed = value;
    }
}

public class Car : Vehicle
{
    private int fuelLevel; // private поле

    public void Refuel(int amount) // public метод
    {
        fuelLevel += amount;
    }

    protected void CheckSpeedLimit() // protected метод
    {
        if (speed > 120)
        {
            Console.WriteLine("Превышен скоростной режим!");
        }
    }
}
```

В этом примере:
- Свойство `Color` в классе `Vehicle` объявлено как `public`, что позволяет доступ к нему из любого места в коде.
- Поле `speed` в классе `Vehicle` объявлено как `private`, что ограничивает доступ к нему только внутри класса `Vehicle`.
- Метод `SetSpeed` в классе `Vehicle` объявлен как `protected`, что позволяет доступ к нему из класса `Vehicle` и его производных классов.
- Поле `fuelLevel` в классе `Car` объявлено как `private`, ограничивая доступ к нему только внутри класса `Car`.
- Методы `Refuel` и `CheckSpeedLimit` в классе `Car` демонстрируют использование модификаторов доступа для контроля над тем, как объекты могут взаимодействовать друг с другом.

### Заключение

Создание экземпляров классов и взаимодействие между объектами являются ключевыми аспектами объектно-ориентированного программирования в C#. Правильное использование модификаторов доступа обеспечивает безопасность данных и инкапсуляцию, позволяя разработчикам создавать более надежные и управляемые программы.

````
public class Vehicle
{
    public string Color { get; set; } 
    public int speed; 

    protected void SetSpeed(int value) 
    {
        speed = value;
    }
}

public class Car : Vehicle
{
    private int fuelLevel;

    public void Refuel(int amount) 
    {
        fuelLevel += amount;
    }

    protected void CheckSpeedLimit()
    {
        if (speed > 120)
        {
            Console.WriteLine("Превышен скоростной режим!");
        }
    }
}
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (2-3 например Dog, Cat и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также статические и обычными атрибуты и метода класса с реализацией взаимодействия между объектами.

````
  // Базовый класс Animal
    public class Animal
    {
        // Приватные поля
        private string _name;
        private int _health; // 0..100

        // Защищённое поле (видно в наследниках)
        protected int energy; // условная энергия (0..100)

        // Публичные свойства (геттеры/сеттеры)
        public string Name
        {
            get => _name;
            set
            {
                if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException("Имя не может быть пустым.");
                _name = value;
            }
        }

        public int Health
        {
            get => _health;
            set
            {
                // Ограничиваем здоровье в диапазоне 0..100
                _health = Math.Max(0, Math.Min(100, value));
            }
        }

        // Статический атрибут — общее количество животных
        public static int TotalAnimals { get; private set; } = 0;

        // Только для чтения (инициализируется в конструкторе)
        public readonly DateTime CreatedAt;

        // Конструктор
        public Animal(string name, int initialHealth = 100, int initialEnergy = 50)
        {
            Name = name;
            Health = initialHealth;
            energy = Math.Max(0, Math.Min(100, initialEnergy));
            CreatedAt = DateTime.Now;
            TotalAnimals++;
        }

        // Виртуальный метод — потомки могут переопределять
        public virtual void Speak()
        {
            Console.WriteLine($"{Name} издает неопределённый звук.");
        }

        // Обычные методы
        public void Eat(int foodUnits)
        {
            if (foodUnits <= 0) throw new ArgumentOutOfRangeException(nameof(foodUnits));
            energy = Math.Min(100, energy + foodUnits * 5); // простая модель
            Health = Math.Min(100, Health + foodUnits * 2);
            Console.WriteLine($"{Name} поел(а) ({foodUnits} ед.) -> здоровье: {Health}, энергия: {energy}");
        }

        public void Move(int meters)
        {
            if (meters < 0) throw new ArgumentOutOfRangeException(nameof(meters));
            int cost = meters / 10;
            energy = Math.Max(0, energy - cost);
            Console.WriteLine($"{Name} прошёл(ла) {meters} м. (энергия -{cost} -> {energy})");
        }

        // Получить урон
        public void ReceiveDamage(int damage, Animal attacker = null)
        {
            if (damage <= 0) return;
            Health = Math.Max(0, Health - damage);
            string by = attacker != null ? $" от {attacker.Name}" : "";
            Console.WriteLine($"{Name} получил(а) {damage} урона{by}. Текущее здоровье: {Health}");
            if (Health == 0)
            {
                Console.WriteLine($"{Name} погиб(ла).");
            }
        }

        // Защищённый метод для изменения энергии в наследниках
        protected void SetEnergy(int value)
        {
            energy = Math.Max(0, Math.Min(100, value));
        }

        // Статический метод
        public static void DisplayTotals()
        {
            Console.WriteLine($"Всего животных: {TotalAnimals}");
        }
    }

    // Класс Dog — наследник Animal
    public class Dog : Animal
    {
        private string _breed;
        private int loyalty; // 0..100

        public string Breed
        {
            get => _breed;
            set => _breed = string.IsNullOrWhiteSpace(value) ? "Unknown" : value;
        }

        public bool IsTrained { get; set; }

        // Статический счетчик собак
        public static int TotalDogs { get; private set; } = 0;

        public Dog(string name, string breed, bool trained = false)
            : base(name, initialHealth: 100, initialEnergy: 60)
        {
            Breed = breed;
            IsTrained = trained;
            loyalty = trained ? 80 : 50;
            TotalDogs++;
        }

        public override void Speak()
        {
            Console.WriteLine($"{Name} (собака) лает: Гав!");
        }

        // Собака преследует другое животное
        public void Chase(Animal target)
        {
            if (target == null) throw new ArgumentNullException(nameof(target));
            if (Health == 0) { Console.WriteLine($"{Name} не может преследовать — он(а) мёртв(а)."); return; }
            if (energy < 10) { Console.WriteLine($"{Name} слишком уставший/ая, чтобы преследовать."); return; }

            Console.WriteLine($"{Name} начинает преследовать {target.Name}.");
            energy -= 10;
            // Если цель — кот и он спрятался (используем проверку IsIndoor/Hide через приведение), наносим меньше урона
            if (target is Cat cat)
            {
                if (cat.IsHidden)
                {
                    Console.WriteLine($"{target.Name} спрятался(ась) — урон минимален.");
                    target.ReceiveDamage(3, this);
                }
                else
                {
                    target.ReceiveDamage(12, this);
                }
            }
            else
            {
                // Для других животных обычный урон
                target.ReceiveDamage(8, this);
            }

            // Понижаем лояльность немного при агрессии (символика)
            loyalty = Math.Max(0, loyalty - 2);
            Console.WriteLine($"{Name}: энергия {energy}, лояльность {loyalty}");
        }

        // Защищённый метод — проверка усталости
        protected void CheckTired()
        {
            if (energy < 15)
                Console.WriteLine($"{Name} устал(а). Рекомендуется отдохнуть или поесть.");
        }

        public static void DisplayTotalDogs()
        {
            Console.WriteLine($"Всего собак: {TotalDogs}");
        }
    }

    // Класс Cat — наследник Animal
    public class Cat : Animal
    {
        private bool _isIndoor;
        public bool IsIndoor
        {
            get => _isIndoor;
            set => _isIndoor = value;
        }

        // Признак того, что кот спрятан (влияет на взаимодействие)
        public bool IsHidden { get; private set; } = false;

        // Статический счетчик котов
        public static int TotalCats { get; private set; } = 0;

        public Cat(string name, bool isIndoor = true)
            : base(name, initialHealth: 90, initialEnergy: 70)
        {
            IsIndoor = isIndoor;
            TotalCats++;
        }

        public override void Speak()
        {
            Console.WriteLine($"{Name} (кот) мяукает: Мяу!");
        }

        // Кот царапает цель
        public void Scratch(Animal target)
        {
            if (target == null) throw new ArgumentNullException(nameof(target));
            if (Health == 0) { Console.WriteLine($"{Name} не может царапать — он(а) мёртв(а)."); return; }
            if (energy < 5) { Console.WriteLine($"{Name} слишком уставший/ая, чтобы царапать."); return; }

            Console.WriteLine($"{Name} царапает {target.Name}!");
            energy -= 5;
            target.ReceiveDamage(6, this);
            // После атаки кот может попытаться спрятаться
            Hide();
        }

        // Кот прячется — временно уменьшает урон от преследования
        public void Hide()
        {
            IsHidden = true;
            Console.WriteLine($"{Name} прячется и становится труднееразличим(а) для преследования.");
            // Можно сбросить через таймер/вызов, но для простоты пользовательский код сам снимет флаг:
            // IsHidden = false; когда нужно.
        }

        public void Reveal()
        {
            IsHidden = false;
            Console.WriteLine($"{Name} выходит из укрытия.");
        }

        public static void DisplayTotalCats()
        {
            Console.WriteLine($"Всего котов: {TotalCats}");
        }
    }
    
````

````
// Создаём животных
Dog rex = new Dog("Рекс", "Овчарка", trained: true);
Cat murka = new Cat("Мурка", isIndoor: true);

// Базовые вызовы
rex.Speak();        // Рекс лает
murka.Speak();      // Мурка мяукает

// Состояние
Console.WriteLine($"{rex.Name}: здоровье {rex.Health}, энергия (скрыто) ");
Console.WriteLine($"{murka.Name}: здоровье {murka.Health}, спрятан: {murka.IsHidden}");

// Взаимодействие: собака преследует кота
rex.Chase(murka);   // наносит урон коту

// Кот царапает собаку в ответ
murka.Scratch(rex);

// Кот прячется и собака снова пытается преследовать
murka.Hide();
rex.Chase(murka);   // меньший урон из-за IsHidden

// Кот выходит из укрытия
murka.Reveal();

// Пополнение энергии и еды
rex.Eat(3);         // увеличит здоровье и энергию
murka.Eat(1);

// Показать общую статистику
Animal.DisplayTotals();
Dog.DisplayTotalDogs();
Cat.DisplayTotalCats();

// Демонстрация коллекции и полиморфизма
 List<Animal> yard = new List<Animal> { rex, murka };
Console.WriteLine("\nПолиморфные вызовы Speak() через список:");
foreach (var a in yard)
    a.Speak();

// Пример ручного "воскрешения" (например лечебный приём)
Console.WriteLine("\nПытаемся восстановить здоровье Рекса на 30:");
rex.Health += 30;
Console.WriteLine($"{rex.Name} текущее здоровье: {rex.Health}");
    
````
